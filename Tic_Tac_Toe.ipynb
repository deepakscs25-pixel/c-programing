```python
import random

board = [" "] * 9
player = "X"

def display(b):
    print(f"\n {b[0]} | {b[1]} | {b[2]} ")
    print("---|---|---")
    print(f" {b[3]} | {b[4]} | {b[5]} ")
    print("---|---|---")
    print(f" {b[6]} | {b[7]} | {b[8]} \n")

def select_empty(b):
    empty = [i for i in range(9) if b[i] == " "]
    return random.choice(empty)

def check_win(b, p):
    wins = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],
        [0, 3, 6], [1, 4, 7], [2, 5, 8],
        [0, 4, 8], [2, 4, 6]
    ]
    for condition in wins:
        if b[condition[0]] == b[condition[1]] == b[condition[2]] == p:
            return True
    return False

print("Tic-Tac-Toe Game")

while " " in board:
    display(board)

    if player == "X":
        position = int(input("Enter position (1-9): ")) - 1
        while position < 0 or position > 8 or board[position] != " ":
            position = int(input("Invalid spot! Enter position (1-9): ")) - 1
        board[position] = "X"
    else:
        print("Computer's turn (O)...")
        position = select_empty(board)
        board[position] = "O"

    if check_win(board, player):
        display(board)
        print(f"Player {player} wins!")
        break

    if player == "X":
        player = "O"
    else:
        player = "X"

else:
    display(board)
    print("Game Draw!")
```
